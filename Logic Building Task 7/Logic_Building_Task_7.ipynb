**Problem 1: Smart Light Controller**

```python
# This class represents a smart light
class SmartLight:
    def __init__(self, name):
        # store the light name and default status as OFF
        self.name = name
        self.status = "OFF"

    # method to turn light ON
    def turn_on(self):
        self.status = "ON"

    # method to turn light OFF
    def turn_off(self):
        self.status = "OFF"

    # method to display current status
    def show_status(self):
        print(f"{self.name} light is {self.status}")


light_name = input("Enter light name: ")
action = input("Enter action (ON/OFF): ").upper()

# object
light = SmartLight(light_name)

# perform action
if action == "ON":
    light.turn_on()
else:
    light.turn_off()

# display status
light.show_status()
```

**Problem 2: Employee ID Card System**

```python
# This class stores basic employee details
class EmployeeIDCard:
    def __init__(self, name, emp_id, department):
        self.name = name
        self.emp_id = emp_id
        self.department = department

    # method to display ID card details
    def show_id_card(self):
        print("Employee ID Card")
        print("Name:", self.name)
        print("ID:", self.emp_id)
        print("Department:", self.department)


employee_name = input("Enter Employee Name: ")
employee_id = input("Enter Employee ID: ")
department = input("Enter Department: ")

emp = EmployeeIDCard(employee_name, employee_id, department)
emp.show_id_card()
```

**Problem 3: Mobile Contact Record**

```python
# This class stores a contact's name and phone number
class Contact:
    def __init__(self, name, phone_number):
        self.name = name
        self.phone_number = phone_number

    # method to display contact info
    def show_contact(self):
        print("Contact Saved")
        print("Name:", self.name)
        print("Phone:", self.phone_number)


contact_name = input("Enter Contact Name: ")
phone = int(input("Enter phone: "))

c = Contact(contact_name, phone)
c.show_contact()
```

**Problem 4: Product Price Tag Generator**

```python
# This class stores product name and price
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    # method to print a formatted price tag
    def show_price_tag(self):
        print("Product:", self.name)
        print("Price: ₹" + str(self.price))


product_name = input("Enter Product Name: ")
product_price = int(input("Enter Product Price: "))

p = Product(product_name, product_price)
p.show_price_tag()
```

**Problem 5: Movie Rating Display System**

```python
# This class stores movie name and rating
class Movie:
    def __init__(self, name, rating):
        self.name = name
        self.rating = rating

    # method to display movie details
    def show_details(self):
        print("Movie:", self.name)
        print("Rating:", self.rating, "/ 5")


movie_name = input("Enter Movie Name: ")
movie_rating = float(input("Enter Movie Rating: "))

m = Movie(movie_name, movie_rating)
m.show_details()
```

**Problem 6: Delivery Address Manager**

```python
# This class stores customer name and delivery address
class DeliveryAddress:
    def __init__(self, customer_name, address):
        self.customer_name = customer_name
        self.address = address

    # method to print delivery details
    def show_delivery_details(self):
        print("Delivery Details")
        print("Customer:", self.customer_name)
        print("Address:", self.address)


customer_name = input("Enter Customer Name: ")
delivery_address = input("Enter Delivery Address: ")

d = DeliveryAddress(customer_name, delivery_address)
d.show_delivery_details()
```
