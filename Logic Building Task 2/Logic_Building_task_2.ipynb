Problem 1 : Unique words in a sentence

```python
# Given sentence
sentence = "python is easy and python is powerful"

# split the sentence into words1
words = sentence.split()

# convert the list of words to a set to get unique words
unique_words = set(words)

# print the count of unique words
print("Number of unique words:", len(unique_words))

# print the unique words in the sentence
print("Unique words in the sentence:", unique_words)
```

problem 2 : Highest Salary from Employee Data

```python
# given input

employees = {
    "Ironman": 75000,
    "Spiderman": 68000,
    "Hulk": 72000
}

highest_employee = None
highest_salary = 0

# logic for finding the employee with the highest salary
for name, salary in employees.items():
    if salary > highest_salary:
        highest_salary = salary
        highest_employee = name

print(f"The employee with the highest salary is {highest_employee} with a salary of {highest_salary}.")
```

Prbolem 3 : Find Maximum and Minimum values


```python
# Given list

numbers = [45,50,60,13,15,39,69,100]

maximum = numbers[0]
minimum = numbers[0]

# logic for finding the maximum and minimum values in the list
for num in numbers:
    if num > maximum:
        maximum = num
    if num < minimum:
        minimum = num

print(f"Numbers in the list: {numbers}")
print(f"Maximum value: {maximum}")
print(f"Minimum value: {minimum}")
```

Problem 5 : Calculate attendance percentage

```python
# list of attendance records where "P" stands for present and "A" stands for absent
attendance = ["P","A","P","P","A","P","P","P","A","P"]

present_count = 0
total_days = len(attendance)

for status in attendance:
    if status == "P":
        present_count += 1

attendance_percentage = (present_count / total_days) * 100

print(f"Total attendance: {total_days}")
print(f"Present count: {present_count}")
print(f"Attendance percentage: {attendance_percentage:.2f}%")
```

Problem 7 : count character frequency

```python
# test text
text = "hello world"

# Empty dictionary to store character counts
char_frequency = {}

# loop through each character in the text string
for char in text:
    if char in char_frequency:
        char_frequency[char] += 1
    else:
        char_frequency[char] = 1

# print the final character frequency dictionary
print("Character frequency:", char_frequency)
```

problem 9 : Check if a key exists in a dictionary

```python
# test dictionary
employee_id = {
    "Ironman": 101,
    "Spiderman": 102,
    "Hulk": 103,
    "Thor": 104
}

# taking input from user to check if the employee name exists in the dictionary
employee_name = input("Enter the employee name: ")

# checking if the employee name exists in the dictionary keys
for name in employee_id.keys():
    if name == employee_name:
        print(f"{employee_name} exists in the dictionary with employee ID: {employee_id[name]}")
        break
else:
    print(f"{employee_name} does not exist in the dictionary.")
```

Problem 10: Calulate average marks

```python
# test marks
marks = [85, 90, 78, 92, 88]

# find the total marks by summing the list
total_marks = sum(marks)

# calculate the number of marks
number_of_subjects = len(marks)

# calculate the average marks
average_marks = total_marks / number_of_subjects

# print the average marks
print("Average marks:", average_marks)
```
