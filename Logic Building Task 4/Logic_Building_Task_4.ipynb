**Problem 1: Mobile Recharge Validation System**

```python
# Function to validate a single recharge attempt
def validate_recharge(amount):
    valid_plans = [199, 299, 399, 599]

    # amount must be >= 50
    if amount < 50:
        print("Invalid amount: Minimum recharge amount is ₹50.")
        return False

    # amount must be one of the valid plans
    if amount not in valid_plans:
        print("Invalid plan: Choose a valid plan (199, 299, 399, 599).")
        return False

    # If both rules pass
    print("Recharge Successful for ₹", amount)
    return True


# Allow user to retry until recharge successful
while True:

    user_input = input("Enter recharge amount (₹): ")

    # Convert to integer safely and carefuly
    try:
        amount = int(user_input)
    except ValueError:
        print("Please enter a valid number.")
        continue

    # Calling validation function
    if validate_recharge(amount):
        break   # exit loop if recharge is valid
    else:
        print("Please try again.\n")
```

**Inventory aReorder Alert System**

```python
# Sample dictionary with products as keys and their stock quantities as values
products = {
    "Pens": 10,
    "Notebooks": 25,
    "Markers": 8,
    "Files": 18
}

# Function to check stock status
def check_inventory(stock_dict):
    for product_name, quantity in stock_dict.items():
        if quantity < 15:
            print(product_name, ": Reorder Alert")
        else:
            print(product_name, ": Stock OK")

# Calling the function
check_inventory(products)
```

**Problem: Student Result Processing System**

```python
# Function that accepts a list of marks of student and returns Pass or Fail
def process_result(marks):
    # Calculate total using a for loop
    total = 0
    for m in marks:
        total += m

    average = total / len(marks)
    print("Average Marks:", average)

    # Check pass/fail status
    if average >= 50:
        return "Pass"
    else:
        return "Fail"

# taking input
student_marks = input("Enter the marks by separated by spaces:")

#  Convert the input string into a list
marks_list = []
for value in student_marks.split():
  marks_list.append(int(value))


# Callng the fuction
result = process_result(marks_list)
print("Result:", result)
```

**Problem 4: Cab Fare Estimator with Retry Options**

```python
# This function will calculate the cab fare for one ride
def calculate_fare(distance_km, is_peak_hour):
    base_fare = 50
    per_km_rate = 12

    # normal fare calculation
    fare = base_fare + (distance_km * per_km_rate)

    # if it is peak hour, add 25% extra
    if is_peak_hour:
        fare = fare * 1.25

    return fare


# We will allow the user to try multiple times using a while loop
while True:


    distance_input = input("Enter distance in km: ")

    # try to convert to float, if not possible ask again
    try:
        distance = float(distance_input)
    except ValueError:
        print("Please enter a valid number for distance.\n")
        continue

    # ask if it is peak hour (yes/no)
    peak_input = input("Is it peak hour? (yes/no): ").strip().lower()

    if peak_input == "yes":
        peak = True
    elif peak_input == "no":
        peak = False
    else:
        print("Please type 'yes' or 'no' only.\n")
        continue

    # call the function to get fare
    total_fare = calculate_fare(distance, peak)

    print("Total Cab Fare: ₹", round(total_fare, 2))

    # ask user if they want to calculate again
    again = input("Do you want to calculate again? (yes/no): ").strip().lower()
    if again != "yes":
        print("Thank you for using the cab fare estimator.")
        break
```

**Problem 5: Employee Attendance Eligibility Checker**

```python
# This function takes a list like ["P", "A", "P"] and returns "Eligible" / "Not Eligible"
def check_attendance_eligibility(attendance_list):
    total_days = len(attendance_list)
    present_days = 0

    # count how many times "P" appears in the list
    for status in attendance_list:
        if status == "P":
            present_days += 1

    # calculate attendance percentage
    attendance_percentage = (present_days / total_days) * 100
    print("Attendance Percentage:", attendance_percentage)

    # check eligibility condition
    if attendance_percentage >= 75:
        return "Eligible"
    else:
        return "Not Eligible"


# Example list
attendance = ["P", "P", "A", "P", "P", "A", "P", "P"]

result = check_attendance_eligibility(attendance)
print("Attendance Status:", result)
```

**Problem 6: Password Strength Checker**

```python
def check_password_strength(password):
    # minimum length = 8
    if len(password) < 8:
        return "Weak"

    # must contain at least one digit
    has_digit = False
    # must contain at least one special character (@#$)
    has_special = False

    for ch in password:
        if ch.isdigit():
            has_digit = True
        if ch in "@#$":
            has_special = True

    if has_digit and has_special:
        return "Strong"
    else:
        return "Weak"

password = input("Enter password: ")
print(check_password_strength(password))
```
