## Task 1: Replace Hardcoded Prompts

```python
# converting hardcoded f-string prompt into a reusable LangChain template
explain_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} in simple terms for beginners"
)

# testing with a sample topic
result = explain_template.format(topic="Artificial Intelligence")
print(result)
```

```python
# reusing the same template with different topics
for t in ["Blockchain", "Cloud Computing", "Data Science"]:
    print(explain_template.format(topic=t))
    print()
```

## Task 2: Multi-Input Prompt System

```python
# template that accepts topic, audience, and tone
multi_input_template = PromptTemplate(
    input_variables=["topic", "audience", "tone"],
    template="Explain {topic} for {audience} in a {tone} tone"
)

# test cases as given0
test_cases = [
    {"topic": "AI", "audience": "beginners", "tone": "friendly"},
    {"topic": "Python", "audience": "kids", "tone": "fun"},
    {"topic": "Deep Learning", "audience": "engineers", "tone": "technical"}
]

for case in test_cases:
    prompt = multi_input_template.format(**case)
    print(prompt)
    print()
```

## Task 3: Prompt Variations Engine

```python
# three different prompt styles for the same topic
teaching_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} clearly step by step"
)

interview_template = PromptTemplate(
    input_variables=["topic"],
    template="Ask 3 questions about {topic}"
)

storytelling_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} as a story"
)

topic = "Machine Learning"

templates = {
    "Teaching": teaching_template,
    "Interview": interview_template,
    "Storytelling": storytelling_template
}

for style, tmpl in templates.items():
    print(style + ":")
    print(tmpl.format(topic=topic))
    print()
```

## Task 4: ChatPromptTemplate System

```python
# system messages for different roles
role_behaviors = {
    "teacher": "You are a helpful teacher who explains concepts in a clear and simple way.",
    "interviewer": "You are a strict interviewer who asks tough technical questions.",
    "motivator": "You are an encouraging motivator who inspires learners to keep going."
}

def build_chat_prompt(role, topic):
    # picking the right system message based on role
    system_msg = role_behaviors.get(role, "You are a helpful assistant.")

    chat_template = ChatPromptTemplate.from_messages([
        ("system", system_msg),
        ("human", "Tell me about {topic}")
    ])

    return chat_template.format_messages(topic=topic)

# testing with different roles
roles_to_test = ["teacher", "interviewer", "motivator"]
topic = "Neural Networks"

for role in roles_to_test:
    messages = build_chat_prompt(role, topic)
    print("Role:", role)
    for msg in messages:
        print(" ", type(msg).__name__ + ":", msg.content)
    print()
```

---
## Task 5: Input Validation Layer

```python
# allowed values
VALID_AUDIENCES = ["beginner", "intermediate", "expert"]
VALID_TONES = ["formal", "casual", "fun"]

def validate_inputs(audience, tone):
    # checking audience
    if audience not in VALID_AUDIENCES:
        print("Warning: '" + audience + "' is not a valid audience. Defaulting to 'beginner'.")
        audience = "beginner"

    # checking tone
    if tone not in VALID_TONES:
        print("Warning: '" + tone + "' is not a valid tone. Defaulting to 'formal'.")
        tone = "formal"

    return audience, tone

# valid inputs
print("Test 1 (valid):")
print(validate_inputs("beginner", "fun"))
print()

# invalid audience
print("Test 2 (invalid audience):")
print(validate_inputs("student", "casual"))
print()

# invalid tone
print("Test 3 (invalid tone):")
print(validate_inputs("expert", "serious"))
print()

# both invalid
print("Test 4 (both invalid):")
print(validate_inputs("pro", "angry"))
```

---
## Task 6: Prompt Generator App

```python
# mapping styles to their templates
style_templates = {
    "teaching": PromptTemplate(
        input_variables=["topic", "audience", "tone"],
        template="Explain {topic} for {audience} in a {tone} teaching style"
    ),
    "interview": PromptTemplate(
        input_variables=["topic", "audience", "tone"],
        template="Ask 3 interview questions about {topic} for {audience} in a {tone} interview style"
    ),
    "storytelling": PromptTemplate(
        input_variables=["topic", "audience", "tone"],
        template="Explain {topic} for {audience} in a {tone} storytelling style"
    )
}

def generate_prompt(topic, audience, tone, style):
    # validate first
    audience, tone = validate_inputs(audience, tone)

    # pick the right template
    if style not in style_templates:
        print("Warning: '" + style + "' is not a valid style. Defaulting to 'teaching'.")
        style = "teaching"

    template = style_templates[style]
    return template.format(topic=topic, audience=audience, tone=tone)

# testing the generator
print(generate_prompt("Neural Networks", "beginner", "fun", "storytelling"))
print()
print(generate_prompt("Python", "expert", "formal", "interview"))
print()
print(generate_prompt("Data Science", "intermediate", "casual", "teaching"))
```

## Task 7: Template Reusability Test

```python
# one single template reused across multiple inputs
reusable_template = PromptTemplate(
    input_variables=["topic", "audience", "tone"],
    template="Explain {topic} for {audience} in a {tone} tone"
)

# five different input sets
inputs = [
    {"topic": "Machine Learning", "audience": "beginner", "tone": "fun"},
    {"topic": "Blockchain", "audience": "expert", "tone": "formal"},
    {"topic": "Cybersecurity", "audience": "intermediate", "tone": "casual"},
    {"topic": "Quantum Computing", "audience": "beginner", "tone": "casual"},
    {"topic": "Cloud Computing", "audience": "expert", "tone": "formal"}
]

print("Same template, different outputs:\n")
for i, inp in enumerate(inputs, 1):
    prompt = reusable_template.format(**inp)
    print("Input " + str(i) + ": " + prompt)
    print()
```

## Pipeline Summary

**User Input → Validation → Prompt Template → Dynamic Prompt Generation → Output**

```python
# full pipeline demo
print("=== Full Pipeline Demo ===")
print()

user_topic = "Neural Networks"
user_audience = "student"
user_tone = "serious"
user_style = "storytelling"

print("Raw Input:")
print("  topic:", user_topic)
print("  audience:", user_audience)
print("  tone:", user_tone)
print("  style:", user_style)
print()

# step 1: validate
print("Step 1 - Validation:")
validated_audience, validated_tone = validate_inputs(user_audience, user_tone)
print("  audience:", validated_audience)
print("  tone:", validated_tone)
print()

# step 2: generate prompt
print("Step 2 - Generated Prompt:")
final_prompt = generate_prompt(user_topic, user_audience, user_tone, user_style)
print(" ", final_prompt)
```
