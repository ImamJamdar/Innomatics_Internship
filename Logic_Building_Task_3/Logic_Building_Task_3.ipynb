**Probelm 1: Employee Performance Bonus Eligibility**

```python
# Given dictionary of employees and their performance scores
employees = {
    "Ravi": 92,
    "Anita": 88,
    "Kiran": 92,
    "Surash": 85
}

# Find the highest performance score
highest_score = max(employees.values())

# Collect all employees who have this highest score
top_performers = []
for name, score in employees.items():
    if score == highest_score:
        top_performers.append(name)

# convert list of names to comma-separated string
names_str = ", ".join(top_performers)

print(f"Top Performers Eligible for Bonus: {names_str} (Score: {highest_score})")

```

**Problem 2: Search Query Keyword Analysis**

```python
# Given search query sentence
query = "Buy mobile phone buy phone online"

# Convert the whole sentence to lowercase
query = query.lower()

# Split the sentence into individual words (keywords)
words = query.split()

# Count frequency of each keyword using a dictionary
keyword_freq = {}

for word in words:
    if word in keyword_freq:
        keyword_freq[word] += 1
    else:
        keyword_freq[word] = 1

# Create a new dictionary with only keywords searched more than once
repeated_keywords = {}

for word, count in keyword_freq.items():
    if count > 1:
        repeated_keywords[word] = count

print(repeated_keywords)
```

**Problem 3: Sensor Data Validation**

```python
# Given list of sensor readings where index = hour, value = reading
sensor_readings = [3,4,7,8,10,12,5]

# List to store valid readings as (hour_index, reading_value) pairs
valid_readings = []

# Loop through the list using index and value together
for hour_index, reading_value in enumerate(sensor_readings):

    if reading_value % 2 == 0:
        valid_readings.append((hour_index, reading_value))

# Print the final valid sensor readings
print("Valid Sensor Readings (Hour, Value):",valid_readings)
```

**Problem 4: Email Domain Usage Analysis**

```python
# Given list of employee email IDs
emails = [
    "ravi@gmail.com",
    "anita@yahoo.com",
    "kiran@gmail.com",
    "suresh@gmail.com",
    "meena@yahoo.com"
]

# Dictionary to store domain → count
domain_counts = {}

# Extract domain from each email and count
for i in emails:
    # Split at '@' and take the part after it as domain
    username, domain = i.split("@")

    if domain in domain_counts:
        domain_counts[domain] += 1
    else:
        domain_counts[domain] = 1


total_emails = len(emails)

# Print percentage usage for each domain
for domain, count in domain_counts.items():
    percentage = (count / total_emails) * 100
    print(f"{domain}: {percentage:.0f}%")
```

**Problem 5: Sales Spike Detection**

```python
# Given list
sales = [1200, 1500, 900, 2200, 1400, 3000]

# Average daily sales calculation
total_sales = sum(sales)
num_days = len(sales)
average_sales = total_sales / num_days

# Finding days with sales above average
threshold = average_sales * 1.3

for i, sale_value in enumerate(sales, start=1):
  if sale_value > threshold:
    print(f"Day {i}: {sale_value}")
```

**Problem 6: Duplicate User ID Detection**

```python
# Given list
user_ids = ["user1", "user2", "user1", "user3", "user1", "user3"]

# Dictionary to store user_id counts
id_counts = {}

# Count how many times each user ID appears
for u_id in user_ids:
    if u_id in id_counts:
        id_counts[u_id] += 1
    else:
        id_counts[u_id] = 1

# Print only those IDs which are duplicates or apeared more than once
for uid, count in id_counts.items():
    if count > 1:
        print(f"{uid} → {count} times")
```
