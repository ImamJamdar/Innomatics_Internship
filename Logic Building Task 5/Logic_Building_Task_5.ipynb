**Problem 1: Social Media – Post Engagement Analyzer**

```python
# This function takes a list of like counts and tells us the total likes and whether the post is Viral or Normal.
def analyze_post_engagement(likes_list):
    total_likes = 0

    for likes in likes_list:
        total_likes += likes

    print("Total Likes:", total_likes)

    # check if it is a viral post or normal engagement
    if total_likes >= 1000:
        print("Post Status: Viral Post")
    else:
        print("Post Status: Normal Engagement")


# Example usage
likes_data = [300, 250, 200, 300]
analyze_post_engagement(likes_data)
```

**Problem 3: Agriculture – Rainfall Adequacy Checker**

```python
# This function takes a list of rainfall values and a required level,
# then prints the average and whether rainfall is adequate or not.
def check_rainfall_adequacy(rainfall_list, required_level):
    total = 0

    # calculate total rainfall using a loop
    for value in rainfall_list:
        total += value

    average = total / len(rainfall_list)

    print("Average Rainfall:", average)

    if average >= required_level:
        print("Rainfall Status: Adequate Rainfall")
    else:
        print("Rainfall Status: Inadequate Rainfall")


rainfall_input = input("Enter rainfall values separated by spaces: ")

# convert input string to list of integers
rainfall_values = []
for x in rainfall_input.split():
    rainfall_values.append(int(x))


required_level = 70 # threshold

# call the function
check_rainfall_adequacy(rainfall_values, required_level)
```

**Problem 4: Social Media – Duplicate Account Detection**

```python
# This function checks if there are any duplicate usernames
def detect_duplicate_accounts(usernames):
    unique_usernames = set(usernames)  # removes duplicates

    if len(unique_usernames) < len(usernames):
        print("Duplicate Accounts Found: Yes")
    else:
        print("Duplicate Accounts Found: No")


# Example list
usernames_list = ["user1", "user2", "user3", "user1"]

detect_duplicate_accounts(usernames_list)
```

**Problem 9: E-Learning – Student Attendance Eligibility System**

```python
# This function calculates attendance % and prints eligibility
def check_exam_eligibility(attendance_list):
    total_days = len(attendance_list)
    present_days = 0

    # count how many days are "P"
    for status in attendance_list:
        if status == "P":
            present_days += 1

    attendance_percentage = (present_days / total_days) * 100

    print("Attendance Percentage:", attendance_percentage)

    if attendance_percentage >= 75:
        print("Exam Eligibility: Eligible")
    else:
        print("Exam Eligibility: Not Eligible")


# Example list (P = Present, A = Absent)
attendance_data = ["P", "P", "A", "P", "P"]

check_exam_eligibility(attendance_data)
```

**Problem 10: Smart Electricity Bill Analyzer**

```python
# This function calculates bill based on units and also tells usage level
def calculate_electricity_bill(units):
    # calculate bill using slabs
    if units <= 100:
        bill = units * 3
    elif units <= 200:
        bill = (100 * 3) + (units - 100) * 5
    else:
        bill = (100 * 3) + (100 * 5) + (units - 200) * 7

    # classify usage based on total bill
    if bill < 500:
        usage = "Low Usage"
    elif bill <= 1500:
        usage = "Moderate Usage"
    else:
        usage = "High Usage"

    return bill, usage


# Example usage
units_consumed = 250
total_bill, usage_status = calculate_electricity_bill(units_consumed)

print("Units Consumed:", units_consumed)
print("Total Bill:", total_bill)
print("Usage Status:", usage_status)
```
