**Problem Statement 1: Smart Parking Lot Management System**

```python
# This function manages parking based on logs and capacity
def manage_parking(parking_capacity, vehicle_logs):
    current_parked = 0
    peak_usage = 0

    # Go through each log entry
    for log in vehicle_logs:
        if log == "IN":
            current_parked += 1
        elif log == "OUT":
            current_parked -= 1

        # Track peak usage
        if current_parked > peak_usage:
            peak_usage = current_parked

    # Print current parked vehicles
    print("Currently Parked Vehicles:", current_parked)

    # Parking status based on capacity
    if current_parked > parking_capacity:
        print("Parking Status: Over Capacity! Alert!")
    elif current_parked == parking_capacity:
        print("Parking Status: Full")
    else:
        print("Parking Status: Available")

    print("Peak Parking Usage:", peak_usage)


# input
capacity = 50
logs = ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"]

manage_parking(capacity, logs)
```

**Problem 2: Online Food Delivery Time Estimator**

```python
# This function estimates delivery time based on distance, traffic and weather
def estimate_delivery_timing(distance_km, traffic_level, weather):
    # base time: assume 5 minutes per km
    time_minutes = distance_km * 5

    # add delay based on traffic
    if traffic_level == "Low":
        time_minutes += 0
    elif traffic_level == "Medium":
        time_minutes += 10
    elif traffic_level == "High":
        time_minutes += 20

    # add delay based on weather
    if weather == "Clear":
        time_minutes += 0
    elif weather == "Rainy":
        time_minutes += 10
    elif weather == "Stormy":
        time_minutes += 20

    print("Estimated Delivery Time:", int(time_minutes), "minutes")

# Taking input from user
distance = int(input("Enter the distance: "))
traffic = input("Enter traffic (low / medium / High): ")
weather_cond = input("Enter weather condition (Clear / Rainy / Stormy): ")

estimate_delivery_timing(distance, traffic, weather_cond)
```

**Problem 3: Movie Theatre Seat Occupancy Analyzer**

```python
# This function calculates occupancy % and prints show status
def analyze_seat_occupancy(total_seats, booked_seats_list):

    booked_count = len(booked_seats_list)
    occupancy = (booked_count / total_seats) * 100

    print("Occupancy:", occupancy, "%")

    # decide show status based on occupancy
    if occupancy == 100:
        print("Show Status: Housefull")
    elif occupancy >= 75:
        print("Show Status: Almost Full")
    else:
        print("Show Status: Seats Available")


# Sample Input
total_seats = 200
booked_seats = [1] * 150

analyze_seat_occupancy(total_seats, booked_seats)
```

**Problem 4: Cloud Server Load Classification System**

```python
# This function calculates average CPU load and prints server status
def classify_server_load(cpu_readings):
    # calculate total using a loop
    total = 0
    for value in cpu_readings:
        total += value

    average = total / len(cpu_readings)

    print("Average CPU Load:", average, "%")

    # classify based on rules
    if average < 50:
        print("Server Status: Normal")
    elif average <= 80:
        print("Server Status: Warning")
    else:
        print("Server Status: Critical")


# Sample Input
readings = [45, 60, 70, 85, 90]

classify_server_load(readings)
```

**Problem 5: Smart Classroom Resource Usage Monitor**

```python
# This function finds which resources are overused
def monitor_resource_usage(usage_dict):
    # we will treat "overused" as usage > 8 hours in a day
    overused = []

    for resource, hours in usage_dict.items():
        if hours > 8:
            overused.append(resource)

    # print overused resources
    if overused:
        print("Overused Resources:", ", ".join(overused))
        print("Energy Alert: Yes")
    else:
        print("Overused Resources: None")
        print("Energy Alert: No")


# Sample Input
resource_usage = {
    "Projector": 6,
    "AC": 9,
    "Lights": 4
}

monitor_resource_usage(resource_usage)
```

**Problem 6: Online Event Registration Capacity Controller**

```python
def manage_event_registrations(event_capacity, total_registrations):
    # confirmed cannot exceed capacity
    if total_registrations <= event_capacity:
        confirmed = total_registrations
        waitlisted = 0
        status = "Open"
    else:
        confirmed = event_capacity
        waitlisted = total_registrations - event_capacity
        status = "Closed"

    print("Confirmed Registrations:", confirmed)
    print("Waitlisted Users:", waitlisted)
    print("Registration Status:", status)


# Sample Input
capacity = 100
registrations = 105

manage_event_registrations(capacity, registrations)
```
