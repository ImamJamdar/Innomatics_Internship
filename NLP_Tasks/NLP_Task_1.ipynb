2. What happens if stopwords are not removed?

   If stopwords are not removed, the text ends up containing a lot of unnecessary words like “is”, “the”, and “and” which don’t add much meaning. Because of this, the important words get less attention, and the model may struggle to focus on what actually matters. It also increases the size of the data, making processing slower and less efficient. Overall, it can reduce the quality and accuracy of the results.


3. Provide two real-world scenarios where removing stopwords can be harmful.

   Removing stopwords can sometimes change the actual meaning of a sentence, especially in sensitive tasks. For example, in sentiment analysis, words like “not” are very important—“I am happy” and “I am not happy” have completely opposite meanings, and removing “not” would give the wrong result. Another case is in question-answering systems, where stopwords help maintain the proper structure of a question; removing them can make the sentence unclear or confusing, affecting the system’s understanding.
  

4. Explain the difference between stemming and lemmatization.

   Stemming and lemmatization are both techniques used to reduce words to their base form, but they work in different ways. Stemming simply cuts off parts of a word without considering its meaning, so it can sometimes produce incorrect or incomplete words (like “studies” becoming “studi”). On the other hand, lemmatization uses proper language rules and context to convert a word into its meaningful base form (like “studies” becoming “study”). In short, stemming is faster but less accurate, while lemmatization is slower but gives more correct and meaningful results.

**Task 2: Build Advanced Preprocessing Function**

```python
import re

def preprocess_text(text):

    # 1. Convert text to lowercase
    text = text.lower()

    # 2. Remove URLs
    text = re.sub(r'http\S+|www\S+', '', text)

    # 3. Remove email-like patterns
    text = re.sub(r'\S+@\S+', '', text)

    # 4. Remove numbers
    text = re.sub(r'\d+', '', text)

    # 5. Handle repeated characters (like "soooo" → "so")
    text = re.sub(r'(.)\1+', r'\1', text)

    # 6. Remove special characters (keep only alphabets and spaces)
    text = re.sub(r'[^a-z\s]', '', text)

    # 7. Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    # 8. Remove very short words (length ≤ 2), but keep "no" and "not"
    tokens = text.split()
    cleaned_tokens = []

    for word in tokens:
        if len(word) > 2 or word in ["no", "not"]:
            cleaned_tokens.append(word)

    # Join tokens back to sentence
    cleaned_sentence = " ".join(cleaned_tokens)

    return cleaned_tokens, cleaned_sentence
```

**Task 3: Stress Testing**

```python
sentences = [
    "Get 100% FREE access now!!!",
    "I absolutely looooved this product 😍😍",
    "Worst service ever... 0/10",
    "Call me at 9876543210",
    "This is THE best course!!!",
    "Visit https://openai.com now!",
    "Nooooo this is baaad!!!",
    "OK OK OK I got it",
    "Win $$$ now!!! Limited offer!!!",
    "I am not happy with this"
]

for sentence in sentences:
    tokens, clean_sentence = preprocess_text(sentence)

    print("Original Text:", sentence)
    print("Cleaned Tokens:", tokens)
    print("Cleaned Sentence:", clean_sentence)
    print("-" * 50)
```

**Task 4: Token Analytics**

```python
for sentence in sentences:
    tokens, clean_sentence = preprocess_text(sentence)

    # Total number of tokens
    total_tokens = len(tokens)

    # Number of unique tokens
    unique_tokens = len(set(tokens))

    # Average token length
    if total_tokens > 0:
        avg_length = sum(len(word) for word in tokens) / total_tokens
    else:
        avg_length = 0

    print("Original Text:", sentence)
    print("Tokens:", tokens)
    print("Total Tokens:", total_tokens)
    print("Unique Tokens:", unique_tokens)
    print("Average Token Length:", round(avg_length, 2))
    print("-" * 50)
```

**Task 5: Frequency Analysis**

```python
from collections import Counter

all_tokens = []

for sentence in sentences:
    tokens, _ = preprocess_text(sentence)
    all_tokens.extend(tokens)


word_freq = Counter(all_tokens)

top_10 = word_freq.most_common(10)

print("Top 10 Most Frequent Words:")
for word, count in top_10:
    print(word, ":", count)


least_5 = word_freq.most_common()[:-6:-1]

print("\nTop 5 Least Frequent Words:")
for word, count in least_5:
    print(word, ":", count)
```

**Task 6: Build Full Pipeline**

```python
def full_pipeline(text_list):

    all_tokens = []
    clean_sentences = []

    for text in text_list:
        tokens, clean_text = preprocess_text(text)

        all_tokens.extend(tokens)
        clean_sentences.append(clean_text)

    return {
        "tokens": all_tokens,
        "clean_sentences": clean_sentences
    }


result = full_pipeline(sentences)

print("All Tokens:", result["tokens"])
print("\nClean Sentences:", result["clean_sentences"])
```

```python
import re

def preprocess_text(text):

    # Handle empty input
    if not text or not isinstance(text, str):
        return [], ""

    text = text.lower()

    # Remove URLs and emails
    text = re.sub(r'http\S+|www\S+', '', text)
    text = re.sub(r'\S+@\S+', '', text)

    # Remove numbers
    text = re.sub(r'\d+', '', text)

    # Handle repeated characters
    text = re.sub(r'(.)\1+', r'\1', text)

    # Remove special characters
    text = re.sub(r'[^a-z\s]', '', text)

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    tokens = text.split()

    # Remove short words (except "no", "not")
    cleaned_tokens = []
    for word in tokens:
        if len(word) > 2 or word in ["no", "not"]:
            cleaned_tokens.append(word)

    cleaned_sentence = " ".join(cleaned_tokens)

    return cleaned_tokens, cleaned_sentence


test_cases = [
    "",                     # empty string
    "😍😍😍😂😂",             # only emojis
    "1234567890",           # only numbers
]

for t in test_cases:
    tokens, clean_text = preprocess_text(t)
    print("Input:", t)
    print("Tokens:", tokens)
    print("Cleaned:", clean_text)
    print("-" * 40)
```
